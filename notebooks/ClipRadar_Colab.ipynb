```python
URL = "https://www.youtube.com/watch?v=..."  #@param {type:"string"}
CLIPS = 5  #@param {type:"slider", min:1, max:10, step:1}
CONTENT = "auto"  #@param ["auto", "gaming", "talk", "sports"]
LAYOUT = "vertical"  #@param ["vertical", "square", "landscape", "auto"]
KEYWORDS = ""  #@param {type:"string"}

from clipradar import Options, run
res = run(Options(source=URL, clips=CLIPS, content=CONTENT, layout=LAYOUT,
                  keywords=[k for k in KEYWORDS.split(',') if k.strip()], out_dir='/content/clips'))
for c in res['clips']:
    print(f"#{c['rank']}  {c['start']:.0f}-{c['end']:.0f}s  score {c['score']:.0f}  {c['title']}  ({c['why']})")
```

```python
from IPython.display import Video, display
import os
for c in res['clips']:
    display(Video(os.path.join(res['dir'], c['file']), embed=True, width=270))
```

```python
import shutil
from google.colab import files
zip_path = shutil.make_archive('/content/clipradar_clips', 'zip', res['dir'])
files.download(zip_path)
```
